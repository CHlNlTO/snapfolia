```python
import os
import shutil

# Define the paths
folder_1 = r"F:\dataset - Copy - Copy\train"
folder_2 = r"F:\dataset - Copy - Copy\val"
folder_for_combination = r"F:\dataset - Copy - Copy\here"

# Function to copy folder structure
def copy_folder_structure(source_folder, destination_folder):
    # List all subfolders (classes) in the source folder
    subfolders = [f for f in os.listdir(source_folder) if os.path.isdir(os.path.join(source_folder, f))]
    
    # Print the folder currently being processed
    print(f"Copying classes from: {source_folder}")
    
    # Loop through each subfolder in the source folder
    total_classes = len(subfolders)
    for idx, subfolder in enumerate(subfolders, start=1):
        # Print the current class with the counter
        print(f"Copying class {idx}/{total_classes}: {subfolder}")
        
        subfolder_path = os.path.join(source_folder, subfolder)
        
        # Create the subfolder in the destination folder
        destination_subfolder = os.path.join(destination_folder, subfolder)
        os.makedirs(destination_subfolder, exist_ok=True)
        
        # Copy the images directly inside each class folder
        for image_file in os.listdir(subfolder_path):
            image_file_path = os.path.join(subfolder_path, image_file)
            if os.path.isfile(image_file_path):
                # Copy each image from the source to the destination subfolder
                shutil.copy(image_file_path, destination_subfolder)

# Copy folder structure from folder_1 and folder_2 to folder_for_combination
copy_folder_structure(folder_1, folder_for_combination)
copy_folder_structure(folder_2, folder_for_combination)

print("Folders and images combined successfully!")
```

```python
import os
import shutil

# Define base paths
base_dir = r"F:\dataset - Copy - Copy"
train_dir = os.path.join(base_dir, "train")
val_dir = os.path.join(base_dir, "val")
output_dir = os.path.join(base_dir, "output")

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

# Get all unique class names from both train and val
train_classes = set(os.listdir(train_dir)) if os.path.exists(train_dir) else set()
val_classes = set(os.listdir(val_dir)) if os.path.exists(val_dir) else set()
all_classes = train_classes.union(val_classes)  # Merge unique class names

# Function to copy images from source to destination
def copy_images(src_images_dir, dest_images_dir):
    if os.path.exists(src_images_dir):  # Check if source directory exists
        os.makedirs(dest_images_dir, exist_ok=True)  # Create destination if needed
        for file in os.listdir(src_images_dir):
            src_file = os.path.join(src_images_dir, file)
            dest_file = os.path.join(dest_images_dir, file)
            if os.path.isfile(src_file) and not os.path.exists(dest_file):  # Avoid duplicates
                shutil.copy2(src_file, dest_file)

# Process each class folder
for class_name in all_classes:
    train_images_path = os.path.join(train_dir, class_name, "images")
    val_images_path = os.path.join(val_dir, class_name, "images")
    output_images_path = os.path.join(output_dir, class_name, "images")

    # Copy images from train and val to output
    copy_images(train_images_path, output_images_path)
    copy_images(val_images_path, output_images_path)

print(f"Images from train and val merged successfully into {output_dir}")
```

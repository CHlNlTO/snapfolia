```python
import os
import shutil
from sklearn.model_selection import train_test_split

def train_val_split_with_labels(data_dir, train_percentage, output_dir):
    """
    Splits the dataset (images and labels) into training and validation sets.

    Parameters:
        data_dir (str): Path to the dataset root directory. Each class should have `images` and `labels` subfolders.
        train_percentage (float): Percentage of the dataset to use for training (e.g., 80 for 80%).
        output_dir (str): Path to save the train/val split folders.

    """
    if not (0 < train_percentage < 100):
        raise ValueError("Train percentage must be between 0 and 100.")

    # Paths for train and val splits
    train_dir = os.path.join(output_dir, "train")
    val_dir = os.path.join(output_dir, "val")
    os.makedirs(train_dir, exist_ok=True)
    os.makedirs(val_dir, exist_ok=True)

    # Iterate through each class folder in the dataset
    for class_name in os.listdir(data_dir):
        class_path = os.path.join(data_dir, class_name)

        # Ensure the directory has `images` and `labels` subfolders
        images_dir = os.path.join(class_path, "images")
        labels_dir = os.path.join(class_path, "labels")

        if not os.path.isdir(images_dir) or not os.path.isdir(labels_dir):
            print(f"Skipping {class_name} as it lacks 'images/' or 'labels/' folders.")
            continue

        print(f"Processing class: {class_name}")

        # Get all image files
        images = [img for img in os.listdir(images_dir) if img.endswith(('.jpg', '.jpeg', '.png'))]

        # Split into train and validation sets
        train_images, val_images = train_test_split(
            images,
            train_size=train_percentage / 100,
            random_state=42
        )

        # Create class subdirectories in train and val directories
        train_class_images_dir = os.path.join(train_dir, class_name, "images")
        train_class_labels_dir = os.path.join(train_dir, class_name, "labels")
        val_class_images_dir = os.path.join(val_dir, class_name, "images")
        val_class_labels_dir = os.path.join(val_dir, class_name, "labels")

        os.makedirs(train_class_images_dir, exist_ok=True)
        os.makedirs(train_class_labels_dir, exist_ok=True)
        os.makedirs(val_class_images_dir, exist_ok=True)
        os.makedirs(val_class_labels_dir, exist_ok=True)

        # Move/copy files for training set
        for image in train_images:
            image_path = os.path.join(images_dir, image)
            label_path = os.path.join(labels_dir, os.path.splitext(image)[0] + ".txt")

            # Copy images and labels
            shutil.copy(image_path, os.path.join(train_class_images_dir, image))
            if os.path.exists(label_path):
                shutil.copy(label_path, os.path.join(train_class_labels_dir, os.path.basename(label_path)))

        # Move/copy files for validation set
        for image in val_images:
            image_path = os.path.join(images_dir, image)
            label_path = os.path.join(labels_dir, os.path.splitext(image)[0] + ".txt")

            # Copy images and labels
            shutil.copy(image_path, os.path.join(val_class_images_dir, image))
            if os.path.exists(label_path):
                shutil.copy(label_path, os.path.join(val_class_labels_dir, os.path.basename(label_path)))

    print(f"Train-val split completed. Check '{output_dir}' for results.")
```

```python
data_dir = "F:/SNAP-DS-4"  
output_dir = "F:/SNAP-DS-4-SPLIT"  

train_percentage = 80

train_val_split_with_labels(data_dir, train_percentage, output_dir)
```

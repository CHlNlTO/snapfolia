```python
import os
import shutil

# Define the paths
folder_1 = r"F:\Output_3.2_Env_and_WBG\train"
folder_2 = r"F:\Output_3.2_Env_and_WBG\val"
folder_for_combination = r"F:\SNAP-DS-36"

# Function to copy folder structure
def copy_folder_structure(source_folder, destination_folder):
    # List all subfolders (classes) in the source folder
    subfolders = [f for f in os.listdir(source_folder) if os.path.isdir(os.path.join(source_folder, f))]
    
    # Print the folder currently being processed
    print(f"Copying classes from: {source_folder}")
    
    # Loop through each subfolder in the source folder
    total_classes = len(subfolders)
    for idx, subfolder in enumerate(subfolders, start=1):
        # Print the current class with the counter
        print(f"Copying class {idx}/{total_classes}: {subfolder}")
        
        subfolder_path = os.path.join(source_folder, subfolder)
        
        # Create the subfolder in the destination folder
        destination_subfolder = os.path.join(destination_folder, subfolder)
        os.makedirs(destination_subfolder, exist_ok=True)
        
        # Copy the images directly inside each class folder
        for image_file in os.listdir(subfolder_path):
            image_file_path = os.path.join(subfolder_path, image_file)
            if os.path.isfile(image_file_path):
                # Copy each image from the source to the destination subfolder
                shutil.copy(image_file_path, destination_subfolder)

# Copy folder structure from folder_1 and folder_2 to folder_for_combination
copy_folder_structure(folder_1, folder_for_combination)
copy_folder_structure(folder_2, folder_for_combination)

print("Folders and images combined successfully!")
```

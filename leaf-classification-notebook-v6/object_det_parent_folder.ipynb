```python
import os
import shutil

# Define source and destination paths
train_images = r"F:\SNAPFOLIA\OBJ_DET_DS_40\train\images"
val_images = r"F:\SNAPFOLIA\OBJ_DET_DS_40\val\images"
train_labels = r"F:\SNAPFOLIA\OBJ_DET_DS_40\train\labels"
val_labels = r"F:\SNAPFOLIA\OBJ_DET_DS_40\val\labels"

merged_images = r"F:\SNAPFOLIA\OBJ_DET_DS_PARENT\images"
merged_labels = r"F:\SNAPFOLIA\OBJ_DET_DS_PARENT\labels"

# Ensure destination directories exist
os.makedirs(merged_images, exist_ok=True)
os.makedirs(merged_labels, exist_ok=True)

# Function to move files
def move_files(src_folder, dst_folder):
    for file_name in os.listdir(src_folder):
        src_path = os.path.join(src_folder, file_name)
        dst_path = os.path.join(dst_folder, file_name)
        shutil.move(src_path, dst_path)

# Move images and labels
move_files(train_images, merged_images)
move_files(val_images, merged_images)
move_files(train_labels, merged_labels)
move_files(val_labels, merged_labels)

print("All images and labels have been merged successfully!")
```

#### Copy Dataset for Experimentation

```python
import os
import shutil
import random

# Paths
snap_dataset = r"D:\Snapfolia Dataset\SNAPFOLIA_DATASET"
experiment_dataset = r"D:\Snapfolia Dataset\5CLASS_150IMG"

# Ensure the destination directory exists
os.makedirs(experiment_dataset, exist_ok=True)

# Select 5 classes randomly or specify them
classes = os.listdir(snap_dataset)
selected_classes = random.sample(classes, 5)  # Randomly select 5 classes

# Copy images
for class_name in selected_classes:
    source_class_path = os.path.join(snap_dataset, class_name)
    destination_class_path = os.path.join(experiment_dataset, class_name)

    os.makedirs(destination_class_path, exist_ok=True)

    # Get list of images in the class folder
    images = os.listdir(source_class_path)
    selected_images = random.sample(images, min(len(images), 150))  # Select up to 150 images

    for image_name in selected_images:
        src = os.path.join(source_class_path, image_name)
        dst = os.path.join(destination_class_path, image_name)
        shutil.copy(src, dst)

print(f"Copied 5 classes with up to 150 images each to {experiment_dataset}.")
```

```python
import matplotlib.pyplot as plt
def count_images_in_directories(base_path):
    directory_counts = {}
    for dirpath, dirnames, filenames in os.walk(base_path):
        if dirpath == base_path:
            # Skip the root directory
            continue
        directory_name = os.path.basename(dirpath)
        image_count = len(filenames)
        directory_counts[directory_name] = image_count
    return directory_counts

image_counts = count_images_in_directories(combined_path)
print(image_counts)
```

```python
def plot_image_counts(base_path, output_file="image_counts.png"):
    # Count images in directories
    image_counts = count_images_in_directories(base_path)
    
    # Sort data for better visualization
    sorted_counts = dict(sorted(image_counts.items(), key=lambda item: item[1], reverse=True))
    
    
    # Extract data for the chart
    directories = list(sorted_counts.keys())
    counts = list(sorted_counts.values())
    
    # Plot horizontal bar chart
    plt.figure(figsize=(10, len(directories) * 0.4))  # Adjust height based on the number of directories
    plt.barh(directories, counts, color='skyblue')
    
    # Add labels and title
    plt.xlabel('Number of Classes')
    plt.ylabel('Classes')
    plt.title(f'Number of Classes {len(image_counts)}')
    
    # Display values on bars
    for index, value in enumerate(counts):
        plt.text(value, index, str(value), va='center')
    
    # Save the plot as PNG
    plt.tight_layout()
    plt.savefig(output_file, format="png", dpi=300)
    plt.close()  # Close the plot to free up memory
    
output_path = "dataset_images/class_count.png"  
plot_image_counts(combined_path, output_path)
print(f"Plot saved as '{output_path}'")
```

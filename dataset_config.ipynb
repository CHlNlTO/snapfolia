```python
# PATHS
import os
import shutil
import matplotlib.pyplot as plt

val_path = r"D:\Snapfolia Dataset\Classification\val"
train_path = r"D:\Snapfolia Dataset\Classification\train"
combined_path = r"D:\Snapfolia Dataset\SNAPFOLIA_DATASET"
```

#### Moving Dataset

```python
def combine_directories(source_paths, destination_path):
    # Ensure the destination path exists
    if not os.path.exists(destination_path):
        os.makedirs(destination_path)

    for source_path in source_paths:
        print(f"Processing source directory: {source_path}")
        for dirpath, dirnames, filenames in os.walk(source_path):
            # Calculate relative path
            relative_path = os.path.relpath(dirpath, source_path)
            combined_subdir = os.path.join(destination_path, relative_path)

            # Ensure subdirectories exist in the destination
            if not os.path.exists(combined_subdir):
                os.makedirs(combined_subdir)

            # Move or copy files to the combined directory
            for file in filenames:
                src_file = os.path.join(dirpath, file)
                dest_file = os.path.join(combined_subdir, file)

                # Log the file movement
                print(f"Moving file: {src_file} to {dest_file}")
                shutil.move(src_file, dest_file)
                # Or use shutil.copy if you want to keep the original file:
                # shutil.copy(src_file, dest_file)
        print("Done Moving Files...")

# Combine train and validation directories
combine_directories([train_path, val_path], combined_path)
```

```python
import os
def walk_through_dir(dir_path):
  for dirpath, dirnames, filenames in os.walk(dir_path):
    print(f"There are {len(dirnames)} directories and {len(filenames)} images in '{dirpath}'.")

walk_through_dir(train_path)
```

#### Visualizing Dataset

```python
def count_images_in_directories(base_path):
    directory_counts = {}
    for dirpath, dirnames, filenames in os.walk(base_path):
        if dirpath == base_path:
            # Skip the root directory
            continue
        directory_name = os.path.basename(dirpath)
        image_count = len(filenames)
        directory_counts[directory_name] = image_count
    return directory_counts

image_counts = count_images_in_directories(combined_path)
print(image_counts)
```

```python
def plot_image_counts(base_path, output_file="image_counts.png"):
    # Count images in directories
    image_counts = count_images_in_directories(base_path)
    
    # Sort data for better visualization
    sorted_counts = dict(sorted(image_counts.items(), key=lambda item: item[1], reverse=True))
    
    
    # Extract data for the chart
    directories = list(sorted_counts.keys())
    counts = list(sorted_counts.values())
    
    # Plot horizontal bar chart
    plt.figure(figsize=(10, len(directories) * 0.4))  # Adjust height based on the number of directories
    plt.barh(directories, counts, color='skyblue')
    
    # Add labels and title
    plt.xlabel('Number of Classes')
    plt.ylabel('Classes')
    plt.title(f'Number of Classes {len(image_counts)}')
    
    # Display values on bars
    for index, value in enumerate(counts):
        plt.text(value, index, str(value), va='center')
    
    # Save the plot as PNG
    plt.tight_layout()
    plt.savefig(output_file, format="png", dpi=300)
    plt.close()  # Close the plot to free up memory
    
output_path = "dataset_images/class_count.png"  
plot_image_counts(combined_path, output_path)
print(f"Plot saved as '{output_path}'")
```

```python
import pandas as pd

# Predefined values
train_start = "2024-12-19 19:37:42.302552"
train_end = "2024-12-20 13:00:23.630685"
train_duration = "17:22:41.328133"
epochs = 50
train_dataset = "46,000"
test_dataset = "12,000"
model_name = "Yolov8"  # Add your model name
model_type = "Object Detection"  # Add your model type

# Extract the date from the train_start and train_end inputs and remove milliseconds
train_start_date, train_start_time = train_start.split(" ", 1)
train_end_date, train_end_time = train_end.split(" ", 1)

train_start_time = train_start_time.split('.')[0]
train_end_time = train_end_time.split('.')[0]

# Remove milliseconds from the train_duration
train_duration = train_duration.split('.')[0]

# Create a dictionary to store the data
data = {
    "model_name": [model_name],
    "model_type": [model_type],
    "train_start_date": [train_start_date],
    "train_start_time": [train_start_time],
    "train_end_date": [train_end_date],
    "train_end_time": [train_end_time],
    "train_duration": [train_duration],
    "epochs": [epochs],
    "train_dataset": [train_dataset],
    "test_dataset": [test_dataset],
}

# Convert the dictionary into a pandas DataFrame
training_metrics_df = pd.DataFrame(data)

# Save the DataFrame to a CSV file
training_metrics_df.to_csv("training_metrics.csv", index=False)

# Print the DataFrame
training_metrics_df
```
